```python
letra = str(input('Digite uma letra (F) ou (M): '))
if letra == 'F' or letra == 'f': #pode-se ser utilizado o if not também
  print('F-Feminino')
elif letra == 'M' or letra == 'm':
  print('M-Masculino')
else:
  print('Entrada inválida')
```

```python
letra = input('digite uma letra: ').lower()

vogais = 'aeiou'

if len(letra) == 1 and letra.isalpha():
  if letra in vogais:
    print(f'a letra {letra} é uma vogal.')
  else:
    print(f'a letra {letra} é uma consoante.')
else:
  print('não entendi[erro]. por favor, digite apenas uma unica letra.')
```

```python
print('INFORME O NOME DOS PRODUTOS')
p1 = str(input('Produto 1: '))
p2 = str(input('Produto 2: '))
p3 = str(input('Produto 3: '))

print('INFORME O VALOR')
v1 = float(input(f'{p1} valor: '))
v2 = float(input(f'{p2} valor: '))
v3 = float(input(f'{p3} valor: '))

if v1 <= v2 and v1 <= v3:
  print(f'Você deve comprar {p1}')
elif v2 <= v1 and v2 <= v3:
  print(f'Você deve comprar {v2}')
elif v3 <= v1 and v3 <= v2:
  print(f'Você deve comprar {v3}')
```

```python
# Leitura dos três números
n1 = float(input("Digite o primeiro número: "))
n2 = float(input("Digite o segundo número: "))
n3 = float(input("Digite o terceiro número: "))

print("Ordem decrescente:", end=" ")
if n1 >= n2 and n1 >= n3:
    if n2 >= n3:
        print(n1, n2, n3)
    else:
        print(n1, n3, n2)
elif n2 >= n1 and n2 >= n3:
    if n1 >= n3:
        print(n2, n1, n3)
    else:
        print(n2, n3, n1)
else:
    if n1 >= n2:
        print(n3, n1, n2)
    else:
        print(n3, n2, n1)
#exercicio feito com apoio devido a não lembrar regras para a fórmula
```
