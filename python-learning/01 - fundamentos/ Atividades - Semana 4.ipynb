```python
dia = int(input('Digite um número de 1 a 7: '))

match dia:
  case 1: print('Domingo')
  case 2: print('Segunda-feira')
  case 3: print('Terça-feira')
  case 4: print('Quarta-feira')
  case 5: print('Quinta-feira')
  case 6: print('Sexta-feira')
  case 7: print('Sabado')
  case _: print('Valor inválido')
```

```python
n1 = int(input('Numero 1: '))
n2 = int(input('Numero 2 : '))
operar = str(input('Selecione("+", "-", *, "/"): '))
resultado = 0
match operar:
  case '+':
    resultado = n1 + n2
  case '-':
    resultado = n1 - n2
  case '*':
    resultado = n1 * n2
  case '/':
    resultado = n1 / n2
print(f'Somando {n1} {operar} {n2} = {resultado}')
```

```python
print('CLASSIFICAÇÃO DE NOTAS')
nota = int(input('Nota do aluno: '))

match nota:
  case 10 | 9:
    print('A')
  case 8 | 7:
    print('B')
  case 6 | 5:
    print('C')
  case _:
    print('R (Reprovado)')
```

```python
print('IDENTIFICADOR DE CORES')
cor = str(input('Cor: '))
match cor.lower():
  case 'vermelho' | 'azul' | 'amarelo':
    print('Primária')
  case _:
    print('Cor diferente')
```

```python
print('VERIFICAÇÃO DE EXTENSÃO DE ARQUIVO')
nome_arquivo = str(input('Nome do arquivo: ')).strip() # Remover espaços em branco
partes_nome = nome_arquivo.split('.') # Buscar o elemento .

if len(partes_nome) > 1:
  extensao = partes_nome[-1].lower()

  match extensao:
    case 'jpg' | 'png':
      print('Imagem')
    case 'mp4' | 'mkv':
      print('Vídeo')
    case 'pdf' | 'docx':
      print('Documento')
    case _:
      print('Extensão desconhecida')
else:
  print('Formato de arquivo inválido ou sem extensão.')
```

```python
def identificar_coordenada(coordenada):
  match coordenada:
    case (0, 0):
      return "Origem"
    case (x, 0) if x != 0:
      return "Eixo X"
    case (0, y) if y != 0:
      return "Eixo Y"
    case (x, y):
      return "Coordenada genérica"


print(identificar_coordenada((0, 0)))      #
print(identificar_coordenada((5, 0)))      #Eixo X
print(identificar_coordenada((-3, 0)))     #  Eixo X
print(identificar_coordenada((0, 7)))      # Eixo Y
print(identificar_coordenada((0, -2)))     # EixY
print(identificar_coordenada((2, 3)))
print(identificar_coordenada((-1, 4)))
```

```python
print('ANALISADOR DE LISTAS')
print('-' * 20)

lista = [2]

tamanho_lista = len(lista)

match tamanho_lista:
  case 0:
    print('vazia')
  case 1:
    print('um elemento')
  case 2:
    print(f'Dois elementos {lista}')
  case _:
    print('Mais de dois elementos')
```

```python
print('Conversor')
print('-' * 20) #Aparência
celc_unid = int(input('Unidade de Celcius: '))

celc_fah = (celc_unid * 1.8) + 32#calculos matematicos para a conversão
celc_kelv = celc_unid + 273,15

print('Conversor de unidades')#Tabela de escolha do ususario
print('1 - Fahrenheit')
print('2 - Kelvin')

escolha_conv = int(input('Unidade de destino: '))#escolha do usuario em numero

match escolha_conv:#condições da esolha
  case 1:
    print(f'Em Fahrenheit: {celc_fah}')
  case 2:
    print(f'Em Kelvin: {celc_kelv}')
  case _:
    print('Dados inválidos')
```

```python
print('MOTOR DE COMANDOS')#aparencia/letras
print('-' * 20)
print('ir norte\nir sul\npegar espada\npegar escudo\nparar')
print('-' * 20)

jogo_rodando = True#inicio de true lopping

while jogo_rodando:
    entrada_user = input('Motor: ').split()#separador das caixinhas de palavra

    acao = entrada_user[0]#caixa 1

    if acao != "parar":#Se ação não for parar vai para linha abaixo
        objeto = entrada_user[1]#caixa 2 do usuario na variavel objeto

    match acao:#o unico problema foi lembrar

        case 'ir':
            if objeto == 'norte':#pesquisei para saber se podia if
                print('Player foi para o norte')

            elif objeto == 'sul':
                print('Player foi para o sul')

        case 'pegar':
            if objeto == 'espada':
                print('Player pegou espada')

            elif objeto == 'escudo':
                print('Player pegou escudo')

        case 'parar':
            print('Jogo encerrado')
            jogo_rodando = False



```
