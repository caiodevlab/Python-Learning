```python

import requests
import time

def verificar_disponibilidade(url):
    """
    Função principal que será o objeto dos nossos testes.
    """
    if not url.startswith('http'):
        url = 'https://' + url

    try:
        inicio = time.time()
        # Definimos um timeout de 5 segundos (Regra de Negócio)
        resposta = requests.get(url, timeout=5)
        fim = time.time()

        return {
            "url": url,
            "status_code": resposta.status_code,
            "tempo_segundos": round(fim - inicio, 4),
            "online": True if resposta.status_code == 200 else False,
            "erro": None
        }
    except requests.exceptions.RequestException as e:
        return {
            "url": url,
            "status_code": None,
            "tempo_segundos": None,
            "online": False,
            "erro": str(type(e).__name__)
        }

# Interface de execução (Usando Formulário do Colab)
#@title Testador de URL { run: "auto" }
url_input = "https://www.google.com" #@param {type:"string"}
resultado = verificar_disponibilidade(url_input)

print(f"\n--- RELATÓRIO DO SISTEMA ---")
print(f"URL Testada: {resultado['url']}")
print(f"Status: {'✅ ONLINE' if resultado['online'] else '❌ OFFLINE'}")
print(f"Código HTTP: {resultado['status_code']}")
print(f"Tempo de Resposta: {resultado['tempo_segundos']}s")
if resultado['erro']:
    print(f"Detalhes do Erro: {resultado['erro']}")
```
