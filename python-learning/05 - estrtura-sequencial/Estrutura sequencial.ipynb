```python
valor = float(input('Quanto vocẽ recebe por hora trabalhada? '))
horas = int(input('Quantas horas você trabalhou? '))
soma = valor * horas
ir = soma * 0.11
inss = soma * 0.08
sind = soma * 0.05
descontos = soma - ir - inss - sind
salar = descontos
print(f'+ Salário Bruto :{soma} R$')
print(f'- IR (11%) :{ir} R$')
print(f'- INSS (8%) :{inss} R$')
print(f'- Sindicato ( 5%) :{sind} R$')
print(f'= Salário Liquido :{salar} R$')
```

```python
area = float(input('Qual o tamanho em metros quadrados da area? '))

litros = (area / 6) * 1.1

#APENAS LATAS DE 18L
latas = int(litros // 18)
if litros % 18 > 0:
    latas += 1
preço = latas * 80

#APENAS GALÕES DE 3,6L
galoes = int(litros // 3.6)
if litros % 3.6 > 0:
    galoes += 1
preçog = galoes * 25

#MISTURA (LATAS + GALÕES
latasm = int(litros // 18)
restante_litros = litros % 18

#galões necessários parao que restou
galoesm = int(restante_litros // 3.6)
if restante_litros % 3.6 > 0:
    galoesm += 1
preçom = (latasm * 80) + (galoesm * 25)

#RESULTADOS
print(f'Comprando apenas latas de 18L Quantidade: {latas} Valor: R$ {preço:.2f}')
print(f'Comprando apenas latas de 3,6L Quantidade: {galoes} Valor: R$ {preçog:.2f}')
print(f'Para menor desperdício Latas: {latasm} | Galões: {galoesm} | Valor: R$ {preçom:.2f}')
#Sor esse exercicio foi dificil devido a não utilização de biblioteca (math) pode-se utilizalas?
```
