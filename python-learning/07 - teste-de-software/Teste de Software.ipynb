```python
# Algoritmo de Decisão de Compra
p1 = float(input("Preço Produto 1: R$ "))
p2 = float(input("Preço Produto 2: R$ "))
p3 = float(input("Preço Produto 3: R$ "))

if p1 < p2 and p1 < p3:
    print("Compre o Produto 1")
elif p2 < p1 and p2 < p3:
    print("Compre o Produto 2")
elif p3 < p1 and p3 < p2:
    print("Compre o Produto 3")
else:
    print("Empate ou valores iguais")
```
